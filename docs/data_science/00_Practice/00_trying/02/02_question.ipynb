## Q01 — 유효한 범주만 남기고 서비스 개수 세기

### 1. `contains('Yes|No')`가 틀리는 이유

`str.contains()`는 부분 문자열을 검색한다. 따라서 `No internet service`에도 `No`가 포함되어 `True`가 된다. 이 문제는 셀 전체 값이 정확히 `Yes` 또는 `No`인지 확인해야 하므로 `isin()`을 사용한다.

```python
cols_q1 = [
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

valid = df_q1[cols_q1].isin(['Yes', 'No']).all(axis=1)
df_q1_valid = df_q1.loc[valid].copy()
```

- `isin(['Yes', 'No'])`: 각 셀을 정확히 비교
- `.all(axis=1)`: 한 행의 6개 값이 모두 유효한지 확인
- `~valid`: 유효하지 않은 행을 선택할 때 사용

특정 값인 `No internet service`만 제거해도 현재 데이터에서는 맞지만, 문제는 **Yes/No가 아닌 모든 범주**를 제거하라고 했으므로 `isin(...).all(axis=1)`이 더 정확하다. 행을 직접 삭제하고 싶다면 `df.drop(index=df.index[~valid])`도 가능하지만 Boolean 필터링이 간결하다.

### 2. `apply()`의 정확한 의미

`DataFrame.apply()`가 항상 차원을 낮추는 것은 아니다. 기본값 `axis=0`에서는 각 열을 Series로 함수에 전달하고, `axis=1`에서는 각 행을 전달한다. 이 문제의 유효성 검사는 `apply()` 없이 `isin()`으로 한 번에 처리할 수 있다.

### 3. 변환, 행 합계, 빈도 계산

```python
df_q1_valid[cols_q1] = (
    df_q1_valid[cols_q1]
```

`replace()` 결과는 다시 대입해야 한다. `value_counts()` 결과는 Series이므로 서비스 개수 라벨은 `.loc[1]`, `.loc[6]`으로 접근하면 의미가 명확하다. 문제는 소수점 **첫째 자리**까지 요구하므로 `round(..., 1)`을 사용한다.

## Q02 — 몫으로 사용 월수를 만들고 전체 상관관계 비교

### 1. `/`가 아니라 `//` 사용

문제에서 `TotalCharges / MonthlyCharges`의 **몫**을 사용하라고 했으므로 일반 나눗셈 `/`가 아니라 몫 연산자 `//`를 사용한다. `/`를 사용한 결과도 반올림하면 우연히 `0.999`가 나오지만 문제 조건을 만족한 풀이는 아니다.

```python
df_q2 = df[['tenure', 'MonthlyCharges']].copy()
df_q2['used_month'] = df['TotalCharges'] // df['MonthlyCharges']
```

### 2. 세 변수의 모든 쌍을 비교하기

`corr()['tenure']`만 선택하면 tenure와 관련된 두 상관계수만 확인한다. 현재 데이터에서는 최댓값이 tenure와 used_month 사이여서 답은 같지만, 문제는 세 변수 간 결과 중 최댓값을 묻기 때문에 전체 상관행렬의 비대각 원소를 비교하는 풀이가 안전하다.

```python
corr_abs = df_q2.corr(method='pearson').abs()
np.fill_diagonal(corr_abs.values, 0)
answer_q2 = round(corr_abs.max().max(), 3)
```

자기 자신과의 상관계수는 항상 1이므로 반드시 제외한다. `np.fill_diagonal()`에는 DataFrame 자체가 아니라 값 배열인 `.values`를 전달한다. 파이썬 숫자는 `abs(x)`, Series/DataFrame은 `.abs()`를 사용할 수 있다.

### 3. 실제 데이터로 `corr_abs.max().max()` 이해하기

`used_month = TotalCharges // MonthlyCharges`로 계산하면 일부 행은 다음과 같다.

| index | tenure | MonthlyCharges | used_month |
|---:|---:|---:|---:|
| 0 | 1 | 29.85 | 1 |
| 1 | 34 | 56.95 | 33 |
| 2 | 2 | 53.85 | 2 |
| 3 | 45 | 42.30 | 43 |
| 7028 | 72 | 103.20 | 71 |
| 7031 | 66 | 105.65 | 64 |

`tenure`와 `used_month` 값이 비슷하게 움직이므로 두 변수의 상관관계가 매우 높을 것으로 예상할 수 있다. 실제 피어슨 상관행렬은 다음과 같다.

|  | tenure | MonthlyCharges | used_month |
|---|---:|---:|---:|
| tenure | 1.000000 | 0.246862 | 0.998831 |
| MonthlyCharges | 0.246862 | 1.000000 | 0.246164 |
| used_month | 0.998831 | 0.246164 | 1.000000 |

행렬은 대칭이므로 `tenure → used_month`와 `used_month → tenure`의 `0.998831`은 같은 관계가 두 번 표시된 것이다. 대각선의 1은 각 변수를 자기 자신과 비교한 결과다.

```python
corr_abs = df_q2.corr(method='pearson').abs()
```

`.abs()`는 음의 상관계수도 양수로 바꾸어 방향이 아닌 상관관계의 강도를 비교하게 한다. 현재 세 상관계수는 모두 양수이므로 실제 숫자는 바뀌지 않는다.

```python
np.fill_diagonal(corr_abs.values, 0)
```

대각선을 0으로 바꾼 실제 행렬:

|  | tenure | MonthlyCharges | used_month |
|---|---:|---:|---:|
| tenure | 0.000000 | 0.246862 | 0.998831 |
| MonthlyCharges | 0.246862 | 0.000000 | 0.246164 |
| used_month | 0.998831 | 0.246164 | 0.000000 |

첫 번째 `.max()`는 각 열의 최댓값을 구한다.

```python
corr_abs.max()

# tenure            0.998831
# MonthlyCharges    0.246862
# used_month        0.998831
```

두 번째 `.max()`는 위 세 값 중 다시 최댓값을 구한다.

```python
corr_abs.max().max()
# 0.998830875...

round(corr_abs.max().max(), 3)
# 0.999
```

즉 `corr()`는 모든 쌍을 계산하지만 `corr()['tenure']`를 선택하면 tenure가 들어간 관계만 남는다. 모든 쌍의 최댓값은 전체 행렬에서 자기상관을 제거한 뒤 `.max().max()`로 구한다. 최댓값에 해당하는 관계는 `tenure ↔ used_month`이다.

## Q03 — 범주 변환, 분할, 정규화, F1-score

### 1. 독립변수 목록을 정확히 옮기기

첫 번째 풀이에는 문제에서 요구한 `StreamingMovies` 대신 `StreamingTV`가 들어갔다. 변수명이 비슷하므로 문제 표를 보고 목록을 그대로 옮긴 뒤 개수와 이름을 다시 확인한다.

```python
cols_X = [
    'SeniorCitizen', 'Partner', 'Dependents', 'tenure',
    'MonthlyCharges', 'TotalCharges', 'OnlineSecurity',
    'OnlineBackup', 'DeviceProtection', 'TechSupport',
    'StreamingMovies', 'PaperlessBilling'
]
```

### 2. Yes/No 외 모든 범주를 일반적으로 -1로 변환하기

```python
def encode_binary(ser):
    if ser.dtype == 'object':
        return ser.map(lambda x: 1 if x == 'Yes' else 0 if x == 'No' else -1)
    return ser

X = df[cols_X].copy().apply(encode_binary)
y = df['Churn'].map({'Yes': 1, 'No': 0})
```

`apply()`에 전달한 함수는 object 열뿐 아니라 숫자 열에서도 반드시 값을 `return`해야 한다. 변환된 결과를 새 변수에 저장한 뒤 실제 모델에도 그 변수를 사용해야 한다. `X = df[cols_X].copy()`처럼 복사하면 값을 바꿀 때 발생하는 `SettingWithCopyWarning`도 방지할 수 있다.

### 3. `y`는 1차원 Series

`cols_y = ['Churn']`과 `df[cols_y]`를 사용하면 y가 2차원 DataFrame이 되어 경고가 발생한다. 단일 종속변수는 `y = df['Churn']`처럼 Series로 만든다.

### 4. 데이터 분할 후 학습 데이터로만 정규화

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=123
)

scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

평가 데이터에 `fit_transform()`을 사용하면 평가 데이터의 정보를 미리 학습하는 데이터 누수가 발생한다. 학습 데이터에만 `fit_transform()`, 평가 데이터에는 같은 scaler의 `transform()`만 사용한다. 문제의 '모든 변수'는 모든 **독립변수**를 의미하며 종속변수 y는 정규화하지 않는다.

### 5. 모델 학습과 평가

```python
model = LogisticRegression()
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)
answer_q3 = round(f1_score(y_test, y_pred), 2)
```

seed 123은 데이터 분할의 `random_state`에 적용한다. 기본 solver에서는 `LogisticRegression(random_state=123)`을 추가해도 결과에 실질적인 차이가 없으므로 기준 답안처럼 생략해도 된다.
