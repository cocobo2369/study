## Q01 — 여러 열의 문자열에서 주사율 찾기

### 1. 문제에서 보장한 형식만 사용하기

문제는 주사율이 2~3자리 숫자 뒤에 `Hz` 단위로 표시된다고 보장한다. `Refresh Rate`까지 항상 존재한다고 가정하면 안 된다. 따라서 `Hz`를 기준으로 정규식을 작성한다.

### 2. 문자열 열을 합칠 때 공백 넣기

```python
df_q1['f-p-s'] = (
    df_q1['Frequency'] + ' ' +
    df_q1['Picture_quality'] + ' ' +
    df_q1['Speaker']
)
```

공백 없이 합치면 서로 다른 열의 마지막 문자와 첫 문자가 붙어 검색이 어려워질 수 있다.

### 3. `isin()`과 `str.contains()`의 차이

- `isin(['60 Hz'])`: 셀의 값 전체가 목록의 값과 같은지 검사한다.
- `str.contains('60 Hz')`: 셀의 문자열 일부에 해당 문자가 포함되어 있는지 검사한다.

이 문제처럼 긴 문자열 내부를 검색할 때는 `str.contains()` 또는 `str.extract()`를 사용한다.

### 4. `extract()`는 괄호 안의 부분만 반환한다

```python
freq = df_q1['f-p-s'].str.extract(
    r'(?P<Frequency>\d{2,3})\s*Hz',
    expand=False
)
answer = (freq == '60').sum()
```

- `\d{2,3}`: 숫자 2~3자리
- `\s*`: 공백 0개 이상
- `Hz`: 숫자 뒤에 반드시 있어야 하는 검색 조건
- `(?P<Frequency>...)`: 추출할 그룹의 열 이름을 `Frequency`로 지정
- `expand=False`: 결과를 1차원 Series로 반환

괄호 안에는 숫자만 있으므로 `Hz`는 검색 조건으로 사용될 뿐 결과에는 포함되지 않는다. `Hz`까지 추출하려면 `r'(?P<Frequency>\d{2,3}\s*Hz)'`처럼 `Hz`도 괄호 안에 넣는다.

`expand`를 생략하면 기본값이 `True`이므로 결과가 1열짜리 DataFrame이다. 이 경우 `(freq == '60').sum()`도 숫자 하나가 아니라 Series를 반환한다. 개수 하나만 필요하다면 `expand=False`가 편하다.

추출된 값은 문자열이므로 숫자 `60`이 아닌 문자열 `'60'`과 비교해야 한다. 숫자로 비교하려면 먼저 `pd.to_numeric()`으로 변환한다.

#### 복사본 풀이에서 확인한 DataFrame 열 이름 처리

복사본에서는 `expand=True`인 기본 설정으로 추출하여 결과가 DataFrame이고 열 이름이 정수 `0`이 되었다. 다음 세 방법 중 하나로 명확하게 처리할 수 있다.

```python
# 방법 1: 이름 있는 캡처 그룹 사용
freq_df = df_q1['f-p-s'].str.extract(r'(?P<freq>\d{2,3})\s*Hz')
answer = freq_df['freq'].eq('60').sum()

# 방법 2: 추출 후 열 이름 변경
freq_df = df_q1['f-p-s'].str.extract(r'(\d{2,3})\s*Hz')
freq_df = freq_df.rename(columns={0: 'freq'})
answer = freq_df['freq'].eq('60').sum()

# 방법 3: 처음부터 Series로 반환
freq = df_q1['f-p-s'].str.extract(r'(\d{2,3})\s*Hz', expand=False)
answer = freq.eq('60').sum()
```

`freq_df.isin(['60']).sum()`도 계산은 되지만 열별 합계를 담은 Series를 반환한다. 최종 답 하나를 구할 때는 특정 열을 선택한 뒤 `.eq('60').sum()`을 쓰는 편이 결과 형태가 분명하다.

### 5. 개수만 구할 때는 `contains()`도 가능하다

```python
answer = df_q1['f-p-s'].str.contains(
    r'(?<!\d)60\s*Hz\b',
    na=False
).sum()
```

단순히 `'60 Hz'`만 검색하면 `60Hz`처럼 공백이 없는 표기를 놓친다. 숫자 경계를 지정하지 않으면 `160 Hz`에도 `60 Hz`가 포함될 수 있으므로 정규식으로 정확한 60Hz를 검사하는 편이 안전하다.

## Q03 — 조건 결합, 결측치, 모델 입력 형태

### 1. `str.contains()`의 OR, AND, NOT

```python
# OR: Pixel 또는 Oper가 포함됨
cond_or = (
    df_q3['channel'].str.contains('Pixel', na=False)
    | df_q3['channel'].str.contains('Oper', na=False)
)

# 정규식 하나로 표현한 OR
cond_or = df_q3['channel'].str.contains(r'Pixel|Oper', na=False)

# AND: Netflix와 Youtube가 모두 포함됨
cond_and = (
    df_q3['channel'].str.contains('Netflix', na=False)
    & df_q3['channel'].str.contains('Youtube', na=False)
)

# NOT: 조건에 해당하지 않음
df_filtered = df_q3.loc[~cond_or].copy()
```

Series 조건에서는 파이썬의 `or`, `and`, `not`이 아니라 `|`, `&`, `~`를 사용하고 각 조건을 괄호로 감싼다. 정규식에서 `4K|8K`의 `|`도 OR이다. AND를 정규식 문자열에 단순히 띄어 써서 표현할 수는 없다.

### 2. 파생변수는 필터링된 DataFrame에서 만들기

기존 코드처럼 원본에서 만든 Series를 필터링된 DataFrame에 넣어도 인덱스 정렬 덕분에 현재 결과는 맞다. 그러나 혼동을 줄이려면 필터링한 `df_q3_2`에서 직접 만드는 것이 명확하다.

```python
df_q3_2['Netflix'] = df_q3_2['channel'].str.contains('Netflix', na=False).astype(int)
df_q3_2['PrimeVideo'] = df_q3_2['channel'].str.contains('Prime Video', na=False).astype(int)
df_q3_2['high_quality'] = df_q3_2['Picture_quality'].str.contains(
    r'4K|8K', na=False
).astype(int)
```

`str.contains()`의 결과는 Boolean이므로 `.astype(int)`를 적용하면 `True → 1`, `False → 0`인 이진 변수가 된다.

### 3. `isna()` 뒤에 사용하는 집계

- `df.isna().sum()`: 열별 결측치 개수
- `df.isna().any()`: 열마다 결측치가 하나라도 있는지 확인
- `df.isna().all()`: 열의 값이 모두 결측치인지 확인
- `df.isna().any(axis=1)`: 행마다 결측치가 하나라도 있는지 확인
- `df.isna().all(axis=1)`: 행의 값이 모두 결측치인지 확인

이 문제에서는 `Ratings`가 0이면 `Reviews / Ratings`의 결과가 유효하지 않아 결측치가 생길 수 있다. 입력변수를 만든 다음 결측치 행을 제거하며, 최종 학습 데이터가 문제에서 제시한 **197개**인지 확인한다.

```python
model.fit(X, y)
```

`df_q3_na[['Stars']]`처럼 대괄호를 두 번 쓰면 `(197, 1)` 형태의 2차원 DataFrame이 된다. 단일 종속변수를 기대하는 모델에 이를 전달하면 `A column-vector y was passed when a 1d array was expected` 경고가 발생한다. 대괄호를 한 번 사용하면 경고가 해결된다. 이미 2차원으로 만들었다면 `y.squeeze()` 또는 `y.values.ravel()`도 가능하다.

### 5. 변수 중요도 확인

```python
importance = pd.Series(
    model.feature_importances_,
    index=cols_X
).sort_values(ascending=False)

display(importance)
display(importance.idxmax(), importance.max())
```

Series를 실제로 정렬하는 메서드는 `sort()`가 아니라 `sort_values()`이다. 가장 중요한 변수의 이름은 `idxmax()`, 중요도 값은 `max()`로 구한다. 이 문제는 예측값을 요구하지 않으므로 `model.predict(X)`는 필요 없다.

## 기억할 핵심

1. 긴 문자열 내부 검색은 `str.contains()`, 일부 값 추출은 `str.extract()`를 사용한다.
2. `extract()`는 캡처 그룹 `()` 안의 값만 반환하며, `expand=False`이면 Series가 된다.
3. Series 조건 결합은 `|`, `&`, `~`를 사용한다.
4. 결측치는 `sum()`으로 개수, `any()`로 하나라도 존재하는지, `all()`로 전부 결측인지 확인한다.
5. 회귀모델의 단일 `y`는 `df['Stars']`처럼 1차원으로 전달한다.
