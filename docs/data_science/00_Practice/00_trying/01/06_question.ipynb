````python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier

df = pd.read_csv("../../dataset/edu_enrollees.csv")
display(df.shape)
display(df.columns)
display(df.dtypes)
````

````python
# - **단계 1:** 분석에 사용하지 않을 `city`, `company_size`, `company_type` 컬럼을 제거하시오. --> 제거 - columns_선택
drop_cols = ['city', 'company_size', 'company_type']
df_drop1 = df.drop(columns = drop_cols).copy()
display(df.shape[1] , df_drop1.shape[1])
display('------------------------------')

# - **단계 2:** 각 문자형 변수에 결측치가 하나라도 존재하는 행은 모두 제거하시오. --> 제거-columns_문자열
cols_obj = df_drop1.select_dtypes('object').columns
df_drop2 = df_drop1.dropna(subset = cols_obj).copy()
display(df_drop2[cols_obj].isna().any())
display('------------------------------')

# - **단계 3:** `experience` 변수 값이 `'>20'` 또는 `'<1'`이 있는 행을 제거하고 `experience` 변수의 유형을 정수로 변환하시오. --> 제거 - 행_선택, 변환_dtype 
cond1 = ~df_drop2['experience'].isin(['>20','<1'])
df_drop3 = df_drop2.loc[cond1, :].copy()
df_drop3['experience'] = df_drop3['experience'].astype(int)
display(df_drop2['experience'].dtype, df_drop3['experience'].dtype)
display('------------------------------')

# - **단계 4:** `last_new_job` 변수의 값이 `'>4'` 또는 `'never'`인 행을 제거하고 해당 변수의 유형을 정수로 변환하시오. --> 제거 - 행_선택, 변환_dtype
cond2 = ~df_drop3['last_new_job'].isin(['>4', 'never'])
df_drop4 = df_drop3.loc[cond2, :].copy()
df_drop4['last_new_job'] = df_drop4['last_new_job'].astype(int)
display(df_drop3['last_new_job'].dtype, df_drop4['last_new_job'].dtype)
display('------------------------------')

df_base = df_drop4.copy()
````


### Q02.

`base`를 사용하여 전배 희망 여부(`target`)에 영향을 주는 변수들을 확인하고자 한다.  
다음 절차에 따라 로지스틱 회귀분석을 수행하고 질문에 답하시오.

#### 단계 1

`gender`, `relevant_experience`, `enrolled_university`, `education_level`, `major_discipline` 변수로부터 더미 변수들을 생성한다.

단, 각 변수로부터 더미 변수를 생성할 때 마지막으로 등장하는 범주는 제외하도록 한다.

여기서 마지막으로 등장하는 범주란, 각 컬럼의 값을 사전 순으로 나열하였을 때 마지막으로 등장하는 값이다.

예를 들어, `col` 변수의 범주가 `['A', 'C', 'B', 'B', 'C', 'A']`의 값을 가진다면 사전 상의 마지막 값인 `C`가 제외된다.

#### 단계 2

단계 1에서 생성한 더미 변수와 `city_development_index`, `experience`, `last_new_job`, `training_hours`, `target`, `Xgrp` 변수를 결합하여 새로운 데이터셋을 구성한다.

- 데이터셋명: `job2`
- 이 데이터셋은 문제 3에서도 활용

이 때, `target`, `Xgrp`를 제외한 데이터셋의 컬럼은 아래 순서에 따르도록 한다.

1. `city_development_index`
2. `experience`
3. `last_new_job`
4. `training_hours`
5. `gender`의 더미 변수
6. `relevant_experience`의 더미 변수
7. `enrolled_university`의 더미 변수
8. `education_level`의 더미 변수
9. `major_discipline`의 더미 변수

#### 단계 3

단계 2에서 구성한 데이터셋 `job2`로 다음 조건에 따라 상수항(`Intercept`)이 포함된 로지스틱 회귀분석을 수행한다.

- 종속 변수: `target`
- 독립 변수(총 16개): `target`과 `Xgrp`를 제외한 나머지 변수
- 회귀식에 포함되는 독립 변수의 순서를 컬럼의 순서와 일치시킨다.

**상수항을 제외한 나머지 변수들에 대한 Odds Ratio 중 가장 큰 값을 기술하시오.**

$$
x_i \text{의 Odds Ratio}
=
\frac{
odds(P(Y=1|x_1,\cdots,x_i+1,\cdots,x_n))
}{
odds(P(Y=1|x_1,\cdots,x_i,\cdots,x_n))
}
$$

※ `LogisticRegression()` 클래스의 인자 `C`는 `100000`, `max_iter=1000`, `solver='liblinear'`으로 지정하시오.  
※ `LogisticRegression()` 클래스의 인자 `random_state`는 `123`으로 지정하시오.  
※ 정답은 소수점 셋째 자리에서 버림하여 둘째 자리까지 출력하시오. `(정답 예시: 0.12)`

---

````python
df_q2 = df_base.copy()

# 단계 1 더미변수 생성
cols_dummy = [
    'gender', 
    'relevant_experience', 
    'enrolled_university', 
    'education_level', 
    'major_discipline' 
    ]


df_q2_dummy_list = []
for col in cols_dummy :
    dummy = pd.get_dummies(df_q2[col], prefix=col) #series -pd.get_dummies-> dataframe
    display(dummy.columns)
    dummy = dummy.iloc[ :, : -1].copy()
    display(dummy.columns)
    display("-------------------")
    df_q2_dummy_list.append(dummy)

df_q2_dummy = pd.concat(df_q2_dummy_list, axis = 1)

# 단계2
cols2 = ['city_development_index', 'experience', 'last_new_job', 'training_hours'] 
cols3 = ['target', 'Xgrp' ]

df_q2_cols2 = df_q2[cols2].copy()
df_q2_cols3 = df_q2[cols3].copy()

job2 = pd.concat([
    df_q2_cols2,
    df_q2_dummy,
    df_q2_cols3
], axis = 1)

display(job2.columns)
````

````python
# 단계 3
#D
X = job2.drop(columns = ['target', 'Xgrp']).copy()
y = job2['target'].copy()
#N

#M
model = LogisticRegression(
    C = 100000,
    max_iter = 1000,
    solver = 'liblinear',
    random_state = 123,
    fit_intercept = True
    )
model.fit(X=X,y=y)
display(model.intercept_, model.coef_[0],model.coef_)
display("-------------------------------------")

odds_ratio = np.exp(model.coef_[0])
display(type(odds_ratio))
display("-------------------------------------")

ser_odds_ratio = pd.Series(odds_ratio, index = X.columns)
display(type(ser_odds_ratio),ser_odds_ratio)
display("-------------------------------------")

ser_odds_ratio_sort = ser_odds_ratio.sort_values(ascending = False)
display(type(ser_odds_ratio_sort),ser_odds_ratio_sort)
display("-------------------------------------")

display(ser_odds_ratio_sort.idxmax(), ser_odds_ratio_sort.max())
#E

#---
display(np.floor(ser_odds_ratio_sort.max()*100)/100)
````

### Q03.

`job2`를 이용하여 전체 데이터를 Train과 Test Set으로 나누고, Train Set으로 학습한 모델을 Test Set에 적용하여 모델을 평가하고자 한다.

다음 절차에 따라 분석을 수행하고 질문에 답하시오.

#### 단계 1

2단계에서 구성한 데이터셋 `job2`에서 `Xgrp` 컬럼의 값이 `'train'`인 경우 Train Set으로, `'test'`인 경우 Test Set으로 정의하여 분할한다.

#### 단계 2

아래 가이드에 따라 Train Set으로 K-NN 분류 모델을 학습하고, 이 모델을 Test Set에 적용한다.

- 종속 변수: 전배 희망 여부(`target`)
- 독립 변수(총 16개): 전배 희망 여부(`target`)와 Train/Test set 구분 변수(`Xgrp`)를 제외한 모든 변수
- Euclidean 거리 기준 가장 가까운 5개 데이터의 전배 희망 여부(`target`)를 활용하여 예측

#### 단계 3

예측 결과를 바탕으로 아래 정의된 지표 **A**를 계산하여 기술하시오.

$$
A =
\frac{
(\# \text{ of true positive}) + (\# \text{ of true negative})
}{
(\# \text{ of total data})
}
$$

※ 정답은 반올림하여 소수점 둘째 자리까지 출력하시오. `(정답 예시: 0.12)`

````python
df_q3 = job2.copy()

#단계1
q3_cond_train = job2['Xgrp'].isin(['train'])
q3_cond_test = job2['Xgrp'].isin(['test'])

train = df_q3.loc[q3_cond_train, :].copy()
test = df_q3.loc[q3_cond_test, : ].copy()
````

````python
#단계2
#D
X_train = train.drop(columns = ['target', 'Xgrp']).copy()
y_train = train['target'].copy()

X_test = test.drop(columns = ['target', 'Xgrp']).copy()
y_test = test['target'].copy()

#N

#M(model,fit,predict)
model = KNeighborsClassifier(
    n_neighbors = 5, 
    metric = 'euclidean'
    )
model.fit(X = X_train, y= y_train)
y_pred = model.predict(X_test)

#E
df_q3_ct = pd.crosstab(y_test, y_pred)
display(df_q3_ct)

total = df_q3_ct.sum().sum()
true_positive = df_q3_ct.iloc[1,1]
true_negative = df_q3_ct.iloc[0,0]
display(total, true_positive, true_negative)
round( (true_positive + true_negative)/total, 2)
````

````python
display(round(df_q3_ct.values.diagonal().sum()/df_q3_ct.values.sum(),2))
````

````python
from sklearn.metrics import accuracy_score
round(accuracy_score(y_true = y_test, y_pred = y_pred), 2)
````
