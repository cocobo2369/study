````python
var_X = ["SeniorCitizen", "Partner", "Dependents", "tenure", "MonthlyCharges", "TotalCharges", "OnlineSecurity", "OnlineBackup", "DeviceProtection", "TechSupport", "StreamingMovies", "PaperlessBilling"].copy()
var_y = ["Churn"]
df_q3 = df[var_X + var_y] #독립, 종속은 같은 df에 넣는 습관이 문제풀이 좋음

def series_convert(x) :
    if x == "Yes" :
        return 1
    if x == "No" :
        return 0
    return -1

def dataframe_convert_object(col) : #return을 반드시
    if col.dtype == "object" :
        return col.apply(series_convert)
    return col #제일중요함, return 안하면 None 출력
df_q3_cond = df_q3.apply(dataframe_convert_object)

#df_q3_cond.isin([-1]).sum() 
#찾고자하는 값이 몇개있는지 세려면 count 가 아닌 sum, isin은 true/false를 리턴
#count는 "결측치가 아닌 개수"를 리턴
X = df_q3[var_X]
y = df_q3[var_y]

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3, random_state= 123)
X_train_scaled = MinMaxScaler().fit_transform(X_train)
````
